# Classificação com kNN

Este exemplo usa uma `Pipeline` para que a normalização seja ajustada somente no treino. Os parâmetros do kNN são escolhidos por validação cruzada e o teste é usado uma única vez, no final.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from exemplo import avaliar_knn
```

```python
X, y = make_classification(
    n_samples=120, n_features=6, n_informative=4, n_redundant=0,
    n_classes=3, random_state=42
)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42
)
resultado = avaliar_knn(X_train, y_train, X_test, y_test, folds=3)
resultado
```
