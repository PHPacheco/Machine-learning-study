# kNN sem vazamento de dados

Exemplo canônico: o scaler permanece dentro da `Pipeline`, `k` é escolhido somente por validação cruzada no treino e o teste é consultado uma única vez.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from ml_studies.knn import evaluate_knn
```

```python
X, y = make_classification(
    n_samples=120, n_features=6, n_informative=4, n_redundant=0,
    n_classes=3, random_state=42
)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42
)
result = evaluate_knn(X_train, y_train, X_test, y_test, cv_folds=3)
result.to_dict()
```
