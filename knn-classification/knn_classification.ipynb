# kNN Classification

This example uses a `Pipeline` so that normalization is fitted only on the training data. Cross-validation selects the kNN hyperparameters, and the test set is used only once for the final evaluation.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from example import evaluate_knn
```

```python
X, y = make_classification(
    n_samples=120, n_features=6, n_informative=4, n_redundant=0,
    n_classes=3, random_state=42
)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42
)
result = evaluate_knn(X_train, y_train, X_test, y_test, folds=3)
result
```
